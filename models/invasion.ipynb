# Invasion analyses

We calculate the probability that an antimutator allele invades under different
reproductive modes.

* Population: $N=2000$ diploid individuals. 
* Fitness loci: $L=50$ loci.
* Deleterious mutation rate of the wildtype: $U_d = 0.02$ per genome per generation.
* Deleterious effect of mutation at a fitness locus: $s_d = -0.07$ in a
  heterozygote.
* Dominance and epistasis: deleterious mutations act additively within loci
  (i.e., are codominant) and multiplicatively between loci (i.e., no epistasis).
* The antimutator allele reduced $U_d$ by $\Delta U_d = -0.005$ in a heterozygote and was
codominant. 
* No other mutator or antimutator mutations were allowed to occur. 
* Sex involved random mating and free recombination between adjacent loci $(r = 0.5)$.
* Facultative sexuals reproduced sexually every $1/\phi$ generations. 
* "Rescaled *r*": $r = \phi / 2$.

## Obligate asexual

**Note:** In asexuals, fixation refers to heterozygotes (i.e., `--ptarget 0.5`).

```python
!python facultative.py --multi --reach --ptarget 0.5 --nproc 10 --ploidy 2 --mode asex --N 2000 --nreps 10000 -b 2e-4 0 0 -L 50 20 20 -s .14 1 .5 -d .5 .5 .5 --r 0
```

## Obligate sexual (rescaled *r*)



```python
for tau in [1, 2, 5, 10, 20, 50, 100, 200, 500, 1000, 2000]:
    rr = 0.5 / tau
    !python facultative.py --multi --reach --ptarget 1 --nproc 10 --ploidy 2 --mode sex --N 2000 --nreps 10000 -b 2e-4 0 0 -L 50 20 20 -s .14 1 .5 -d .5 .5 .5 --r {rr}
```
